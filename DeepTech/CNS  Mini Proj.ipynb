````python
# Def. helper funct.
def fe_fingerprint(smi_list, name='morgan'):                    # 定义fe_fingerprint函数，输入为smi_list和name，默认name为'morgan'
    fp_list = []                                                # 初始化一个空列表fp_list
    for smi in smi_list:                                        # 遍历输入列表smi_list
        mol = Chem.MolFromSmiles(smi)                           # 将输入列表中的smi转换为mol格式
        fp = GetMorganFingerprintAsBitVect(mol, 5, nBits=1024)  # 计算mol的Morgan指纹，并将其转换为位向量
        fp = fp.ToBitString()                                   # 将位向量转换为比特字符串
        fp_list.append([int(item) for item in list(fp)])        # 将比特字符串转换为整数列表，并添加到fp_list
    return fp_list                                              # 返回fp_list

def sigmoid(x):                                                 # 定义sigmoid函数，输入为x
    return 1/(np.exp(-x)+1)                                     # 计算并返回sigmoid函数值
````

## 3.3.3 <a id='toc1_3_3'></a>[Baseline Prep.](#toc0_)

本次赛题采用 `F2-score` 作为评价指标。
`F2-score` 是 `F1-beta` 取 $beta = 2$ 时的特殊形式。  `F1-beta` 允许我们调整精确率（Precision）和召回率（Recall）之间的权重，其一般形式如下：

$$F1-beta = (1 + beta^2) * (Precision * Recall) / (beta^2 * Precision + Recall)$$

在本次赛事中，我们更在意模型的召回率，取 $beta = 2$。此时我们得到 `F2-score`，`F2-score` 的计算公式为：

$$F2-score = 5 * (Precision * Recall) / (4 * Precision + Recall)$$

其中，Precision = 真正例（TP）/（真正例（TP）+假正例（FP）），Recall = 真正例（TP）/（真正例（TP）+假负例（FN））。
TP（True Positive）表示实际为正例且预测为正例的样本数；TN（True Negative）表示实际为负例且预测为负例的样本数；FP（False Positive）表示实际为负例但预测为正例的样本数；FN（False Negative）表示实际为正例但预测为负例的样本数。

最终排行榜展示结果只保留 `F2-score` 的四位小数。

## 3.3.4 <a id='toc1_3_4'></a>[Tran & Pred.](#toc0_)

````python
### 3.1 自定义数据预处理与 collate 函数 ###
def custom_molecule_preprocessor(smiles):
    """
    将 SMILES 转换为 Uni-Mol 输入格式（原子和 3D 坐标）
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None
    mol = Chem.AddHs(mol)
    if AllChem.EmbedMolecule(mol, randomSeed=42) != 0:
        return None
    AllChem.MMFFOptimizeMolecule(mol)
    atoms = [atom.GetSymbol() for atom in mol.GetAtoms()]
    coords = mol.GetConformer().GetPositions()
    return {
        'atoms': atoms,
        'coordinates': coords.tolist()
    }

def custom_collate_fn(batch):
    # 简单返回样本列表；实际应用中可能需要对不同样本进行 padding 或其他处理
    # 同时过滤掉预处理失败的样本（返回 None 的情况）
    batch = [sample for sample in batch if sample is not None]
    return batch
````

````python
### 3.2 构建数据集与 DataLoader ###
class UniMolDataset(Dataset):
    def __init__(self, smiles_list, labels=None, preprocessor=None):
        self.smiles_list = smiles_list
        self.labels = labels  # labels 为 None 时表示测试集
        self.preprocessor = preprocessor or custom_molecule_preprocessor
    
    def __len__(self):
        return len(self.smiles_list)
    
    def __getitem__(self, idx):
        smi = self.smiles_list[idx]
        data = self.preprocessor(smi)
        if data is None:
            # 如果预处理失败，返回一个空字典（后续可过滤掉）
            data = {}
        if self.labels is not None:
            data['label'] = self.labels[idx]
        return data

# 使用之前加载的 df_train, df_val, df_test 数据（假设这些 DataFrame 已经存在）
train_dataset_um = UniMolDataset(df_train['SMILES'].tolist(), df_train['TARGET'].tolist())
val_dataset_um   = UniMolDataset(df_val['SMILES'].tolist(),   df_val['TARGET'].tolist())
test_dataset_um  = UniMolDataset(df_test['SMILES'].tolist(),  labels=None)

train_loader_um  = DataLoader(train_dataset_um, batch_size=32, shuffle=True,  collate_fn=custom_collate_fn)
val_loader_um    = DataLoader(val_dataset_um,   batch_size=32, shuffle=False, collate_fn=custom_collate_fn)
test_loader_um   = DataLoader(test_dataset_um,  batch_size=32, shuffle=False, collate_fn=custom_collate_fn)

````

````python
### 3.4.3 加载预训练的 Uni-Mol 模型 ###

device        = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 指定配置文件路径
config_path    = "/opt/conda/lib/python3.8/site-packages/unimol-0.0.2-py3.8.egg/unimol/configs/mol_pre_all_h_220816_config.json"
# 实例化模型
from unimol.models.unimol import UniMolModel
pretrained_model = UniMolModel(config_path=config_path)
# 加载预训练权重
checkpoint = torch.load("/opt/conda/lib/python3.8/site-packages/unimol-0.0.2-py3.8.egg/unimol/weights/mol_pre_all_h_220816.pt", map_location=device)
if "model" in checkpoint:
    checkpoint = checkpoint["model"]

# 构造一个新的状态字典，只保留 backbone 部分的参数
new_state_dict = {}
for k, v in checkpoint.items():
    if k.startswith("lm_head") or k.startswith("pair2coord_proj") or k.startswith("dist_head") or k.startswith("encoder.final_head_layer_norm"):
        continue
    else:
        new_state_dict[k] = v

pretrained_model.load_state_dict(new_state_dict, strict=False)
pretrained_model.to(device)
pretrained_model.eval()

# 定义辅助函数：计算 pairwise 距离矩阵
def compute_distance(coords):
    # coords: tensor, shape (N, 3)
    diff = coords.unsqueeze(0) - coords.unsqueeze(1)  # (N, N, 3)
    distance = diff.pow(2).sum(-1).sqrt()  # (N, N)
    return distance

# 定义辅助函数：构造边类型
ATOM_DICT = {"H": 0, "C": 1, "N": 2, "O": 3, "F": 4, "P": 5, "S": 6, "Cl": 7, "Br": 8, "I": 9}
def get_edge_type(atoms):
    # 这里简单将每个原子符号转换为对应的整数，构造一个 (N, N) 的矩阵
    atom_indices = [ATOM_DICT.get(a, 0) for a in atoms]
    # 简单构造边类型：可以设定为两两原子索引的差的绝对值作为一个示例
    N = len(atom_indices)
    edge_type = torch.zeros((N, N), dtype=torch.long)
    for i in range(N):
        for j in range(N):
            edge_type[i, j] = abs(atom_indices[i] - atom_indices[j])
    return edge_type
````

````python
### 3.4.4 定义基于 Uni-Mol 的性质预测模型 ###

# 定义一个简单的原子符号到索引的映射字典
ATOM_DICT = {"H": 0, "C": 1, "N": 2, "O": 3, "F": 4, "P": 5, "S": 6, "Cl": 7, "Br": 8, "I": 9}

# 定义预测模型
class UniMolPropertyPredictor(torch.nn.Module):
    def __init__(self, unimol_model):
        super().__init__()
        self.unimol = unimol_model
        # 修改这里：如果模型的实际全局表示维度为 512，则设置 d=512
        d = self.unimol.config.hidden_size if hasattr(self.unimol, 'config') else 512
        self.classifier = torch.nn.Linear(d, 1)
    
    def forward(self, batch):
        reps = []
        for sample in batch:
            if 'atoms' not in sample or 'coordinates' not in sample:
                continue
            atoms = sample['atoms']  # list of str
            coords = torch.tensor(sample['coordinates'], dtype=torch.float32).to(device)  # (N, 3)
            src_distance = compute_distance(coords).unsqueeze(0)  # (1, N, N)
            src_coord = coords.unsqueeze(0)  # (1, N, 3)
            src_edge_type = get_edge_type(atoms).to(device).unsqueeze(0)  # (1, N, N)
            src_atom_type = torch.tensor([ATOM_DICT.get(a, 0) for a in atoms], dtype=torch.long).to(device).unsqueeze(0)  # (1, N)
            # 调用预训练模型并传入 return_repr=True
            output = self.unimol(src_atom_type, src_distance, src_coord, src_edge_type, return_repr=True)
            rep = output["cls_repr"]  # 假设全局表示保存在 "cls_repr" 键中，形状 (1, 512)
            reps.append(rep.squeeze(0))
        if len(reps) == 0:
            return torch.empty(0).to(device)
        reps = torch.stack(reps, dim=0)  # (batch_size, 512)
        logits = self.classifier(reps)
        return logits


# 部署+实例化
property_model = UniMolPropertyPredictor(pretrained_model).to(device)
````

````python
### 3.4.5 训练与验证 ###

optimizer_um = torch.optim.Adam(property_model.parameters(), lr=1e-4)
criterion = torch.nn.BCEWithLogitsLoss()

num_epochs = 10
for epoch in range(num_epochs):
    property_model.train()
    train_loss = 0.0
    
    # 使用 tqdm 包装训练 DataLoader
    for batch in tqdm(train_loader_um, desc=f"Epoch {epoch+1}/{num_epochs} - Training"):
        for sample in batch:
            for key in sample:
                if isinstance(sample[key], torch.Tensor):
                    sample[key] = sample[key].to(device)
        optimizer_um.zero_grad()
        logits   = property_model(batch)
        labels   = torch.tensor([sample['label'] for sample in batch], dtype=torch.float32).to(device)
        loss     = criterion(logits.squeeze(), labels)
        loss.backward()
        optimizer_um.step()
        train_loss += loss.item()
    avg_train_loss  = train_loss / len(train_loader_um)
    print(f"Epoch {epoch+1}/{num_epochs} - Train Loss: {avg_train_loss:.4f}")
    
    # 验证阶段
    property_model.eval()
    preds = []
    targets = []
    for batch in tqdm(val_loader_um, desc=f"Epoch {epoch+1}/{num_epochs} - Validation"):
        for sample in batch:
            for key in sample:
                if isinstance(sample[key], torch.Tensor):
                    sample[key] = sample[key].to(device)
        logits = property_model(batch)
        prob = torch.sigmoid(logits.squeeze()).detach().cpu().numpy()
        preds.extend(prob.tolist())
        targets.extend([sample['label'] for sample in batch])
    pred_labels = [1 if p > 0.5 else 0 for p in preds]
    f2 = fbeta_score(targets, pred_labels, beta=2)
    print(f"Epoch {epoch+1} - Validation F2 Score: {f2:.4f}")


````

````python
### 3.6 测试集预测 ###
property_model.eval()
test_preds = []
with torch.no_grad():
    for batch in test_loader_um:
        for sample in batch:
            for key in sample:
                if isinstance(sample[key], torch.Tensor):
                    sample[key] = sample[key].to(device)
        logits = property_model(batch)
        prob = torch.sigmoid(logits.squeeze()).cpu().numpy()
        test_preds.extend(prob.tolist())
````
