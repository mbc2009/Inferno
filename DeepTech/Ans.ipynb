```markdown
# Uni-Mol预训练的做法和策略

## 1. 数据准备
- **分子预训练模型**：使用2.09亿分子构象的数据集进行训练。
- **口袋预训练模型**：基于300万候选蛋白口袋数据进行训练。

## 2. 模型架构
- 两个模型均采用相同的SE(3)-等变变压器架构。

## 3. 预训练任务
- **分子性质预测任务**：通过预训练，模型在MoleculeNet基准的15个数据集中的14个数据集上超过现有最佳方法。
- **口袋性质预测任务**：在包含164,586个候选口袋的数据集上进行回归任务的实验，预训练的Uni-Mol模型在所有四个评分上优于未预训练的模型。
- **蛋白 - 配体结合位点预测任务**：结合分子和口袋预训练模型来学习基于距离矩阵的评分函数，然后对复合物构象进行采样和优化。

## 4. 预训练效果
- 在15个分子性质预测任务中的14个任务上优于现有最佳方法。
- 在CASF - 2016基准测试的对接能力测试中，对于蛋白 - 配体复合物的位点预测，预测出88.07%的结合位点RMSD <= 2Å，比流行的对接方法高出22.81%。
- 在GEOM - QM9和GEOM - Drugs上的分子构象生成任务中，在覆盖率和匹配度指标上达到最佳水平。
- 即使没有预训练，Uni-Mol（称为Uni-Mol random）也优于其他基线方法，这证明了Uni-Mol骨干的有效性。
```
